````python
df = pd.DataFrame(
    {"age": [20, 25, 30],
     "city": ["Almaty", "Berlin", "Furtwangen"]},
    index=["anna", "ben", "carl"]
)

print(df.loc["ben"])         # по названию
print(df.iloc[1])            # по номеру (та же строка)
print(df.loc["anna":"ben"])  # срез по названиям, конец включён
print(df.iloc[0:2])
````

Внутри [] у iloc всегда два места: [строки, колонки], через запятую.

````python
# changing values

data.iloc[:,1]= 3
data.iloc[1,0]= 7
data.iloc[3,1]= 8
data
````

Внутри `[]` у `iloc` всегда два места: **`[строки, колонки]`**, через запятую.

**`df.iloc[:, 1] = 3`**
- `:` перед запятой значит «все строки»
- `1` после запятой значит «колонка на позиции 1» (то есть `b`, счёт с нуля: a=0, b=1, c=2)
- `= 3` записывает 3 во всю колонку `b`

**`df.iloc[1, 0] = 7`**
- строка на позиции 1 (вторая)
- колонка на позиции 0 (`a`)
- одна ячейка, туда пишется 7

Было → стало:
```
   a  b   c           a  b   c
0  1  2  hi        0  1  3  hi
1  3  4  ho   →    1  7  3  ho
2  5  6  ha        2  5  3  ha
```
Колонка `b` стала вся тройками, а `a` во второй строке стала 7. Это те самые два красных выделения на слайде.

**Как читать `[a, b]`:** слева от запятой что выбрать по строкам, справа что по колонкам.

| Запись | Что выбирает |
|---|---|
| `iloc[1, 0]` | одна ячейка |
| `iloc[:, 1]` | вся колонка 1 |
| `iloc[1, :]` | вся строка 1 |
| `iloc[0:2, 1]` | строки 0 и 1 в колонке 1 |

`:` один без чисел = «всё». Это как раз то, что тебя, скорее всего, и сбило.

Тот же результат через `loc`, но по названиям:
```python
df.loc[:, "b"] = 3
df.loc[1, "a"] = 7
```
Здесь индекс строк совпадает с позицией (0, 1, 2), поэтому `1` работает одинаково.

````python
data2 = data.copy()
data2.iloc[2,2] = "new"
data2
````

**Итог по `loc` / `iloc` / index**

- **Index** = названия строк. **Columns** = названия колонок. Это ярлыки, а не сами данные.
- **`loc[строки, колонки]`** работает по **названиям**. Срез включает конец: `loc["a":"c"]`.
- **`iloc[строки, колонки]`** работает по **номерам позиций** (с нуля). Срез не включает конец: `iloc[0:2]`.
- Формат всегда `[строки, колонки]`, а `:` один значит «всё».
- Если индекс не задан, pandas ставит `0, 1, 2...`, и тогда `loc[1]` и `iloc[1]` дают то же самое. После `set_index`, сортировки или фильтрации они расходятся.
- `set_index("col")` превращает колонку в индекс, `reset_index()` возвращает обратно.

```python
df.loc["ben", "city"]    # по названиям
df.iloc[1, 1]            # по номерам (та же ячейка)
df.loc[:, "age"] = 0     # записать во всю колонку
```

**Правило:** есть осмысленные названия, бери `loc`. Нужна позиция, бери `iloc`.
